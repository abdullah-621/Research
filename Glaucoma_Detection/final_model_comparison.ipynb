# Final Model Comparison — Crop Disease Classification

This notebook reads the saved `*_results.json` files produced by each of the five
single-model notebooks (EfficientNetB3, DenseNet121, ResNet50, Xception, ConvNeXtTiny),
builds a comparison table, and selects the **top 3 models** for the next phase
(Confusion Matrix, ROC/AUC, Grad-CAM, SHAP/LIME, and other detailed analysis).

> Run this notebook only **after** all five model notebooks have finished training
> and each has saved its `RESULTS_DIR/<model>_results.json` file.


## 1. Import Libraries

```python
import os
import glob
import json
import pandas as pd
```

## 2. Load Saved Results

```python
RESULTS_ROOT = "Results"   # must match RESULTS_ROOT used in the model notebooks (Results/<MODEL_NAME>/)

result_files = glob.glob(os.path.join(RESULTS_ROOT, "*", "*_results.json"))
print(f"Found {len(result_files)} result file(s):")
for f in result_files:
    print(" -", f)

all_results = []
for file_path in result_files:
    with open(file_path, "r") as f:
        all_results.append(json.load(f))
```

## 3. Build the Comparison Table

```python
df = pd.DataFrame(all_results)

# Reorder / rename columns to match the required reporting format
df = df.rename(columns={
    "model": "Model",
    "train_accuracy": "Train Accuracy",
    "val_accuracy": "Validation Accuracy",
    "test_accuracy": "Test Accuracy",
    "precision": "Precision",
    "recall": "Recall",
    "f1_score": "F1-score",
    "test_loss": "Test Loss",
})[[
    "Model", "Train Accuracy", "Validation Accuracy", "Test Accuracy",
    "Precision", "Recall", "F1-score", "Test Loss",
]]

# Sort by Test Accuracy, highest to lowest
df_sorted = df.sort_values(by="Test Accuracy", ascending=False).reset_index(drop=True)
df_sorted
```

## 4. Save the Comparison Table

```python
comparison_path = os.path.join(RESULTS_ROOT, "final_model_comparison.csv")
df_sorted.to_csv(comparison_path, index=False)
print(f"Comparison table saved to: {comparison_path}")
```

## 5. Select the Top 3 Models

These three models advance to the next phase: Confusion Matrix, ROC Curve, AUC, Grad-CAM, SHAP/LIME, and other detailed explainability analysis.

```python
top3 = df_sorted.head(3)
print("Top 3 models selected for detailed analysis:")
top3
```
