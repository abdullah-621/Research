## 📓 Part-3 — Evaluation, Visualization & Metrics

### Cell 29 — Import Evaluation Metrics

````python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_curve,
    auc,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
````

### Cell 30 — Performance Metrics

````python
# ==========================================================
# PERFORMANCE METRICS
# ==========================================================

def calculate_metrics(labels, predictions, probabilities):

    accuracy = accuracy_score(labels, predictions)

    precision = precision_score(labels, predictions)

    recall = recall_score(labels, predictions)

    f1 = f1_score(labels, predictions)

    fpr, tpr, _ = roc_curve(labels, probabilities)

    roc_auc = auc(fpr, tpr)

    metrics = {

        "Accuracy": accuracy,

        "Precision": precision,

        "Recall": recall,

        "F1 Score": f1,

        "ROC AUC": roc_auc

    }

    return metrics
````

### Cell 31 — Print Metrics

````python
def print_metrics(metrics):

    print("="*50)

    for key, value in metrics.items():

        print(f"{key:<15}: {value:.4f}")

    print("="*50)
````

### Cell 32 — Classification Report

````python
def show_classification_report(labels, predictions):

    print("="*70)

    print("Classification Report")

    print("="*70)

    print(

        classification_report(

            labels,

            predictions,

            target_names=CLASS_NAMES

        )

    )
````

### Cell 33 — Confusion Matrix

````python
def plot_confusion_matrix(labels, predictions):

    cm = confusion_matrix(labels, predictions)

    plt.figure(figsize=(6,5))

    sns.heatmap(

        cm,

        annot=True,

        fmt="d",

        cmap="Blues",

        xticklabels=CLASS_NAMES,

        yticklabels=CLASS_NAMES

    )

    plt.xlabel("Predicted")

    plt.ylabel("Actual")

    plt.title("Confusion Matrix")

    plt.tight_layout()

    plt.show()
````

### Cell 34 — ROC Curve

````python
def plot_roc_curve(labels, probabilities):

    fpr, tpr, _ = roc_curve(

        labels,

        probabilities

    )

    roc_auc = auc(fpr, tpr)

    plt.figure(figsize=(6,6))

    plt.plot(

        fpr,

        tpr,

        linewidth=2,

        label=f"AUC = {roc_auc:.4f}"

    )

    plt.plot(

        [0,1],

        [0,1],

        "--"

    )

    plt.xlabel("False Positive Rate")

    plt.ylabel("True Positive Rate")

    plt.title("ROC Curve")

    plt.legend()

    plt.grid(True)

    plt.show()
````

### Cell 35 — Accuracy Curve

````python
def plot_accuracy(history):

    plt.figure(figsize=(8,5))

    plt.plot(

        history.history["accuracy"],

        label="Training"

    )

    plt.plot(

        history.history["val_accuracy"],

        label="Validation"

    )

    plt.xlabel("Epoch")

    plt.ylabel("Accuracy")

    plt.title("Training vs Validation Accuracy")

    plt.legend()

    plt.grid(True)

    plt.show()
````

### Cell 36 — Loss Curve

````python
def plot_loss(history):

    plt.figure(figsize=(8,5))

    plt.plot(

        history.history["loss"],

        label="Training"

    )

    plt.plot(

        history.history["val_loss"],

        label="Validation"

    )

    plt.xlabel("Epoch")

    plt.ylabel("Loss")

    plt.title("Training vs Validation Loss")

    plt.legend()

    plt.grid(True)

    plt.show()
````

### Cell 37 — Precision & Recall Curve

````python
def plot_precision_recall(history):

    plt.figure(figsize=(8,5))

    plt.plot(

        history.history["precision"],

        label="Train Precision"

    )

    plt.plot(

        history.history["val_precision"],

        label="Validation Precision"

    )

    plt.plot(

        history.history["recall"],

        label="Train Recall"

    )

    plt.plot(

        history.history["val_recall"],

        label="Validation Recall"

    )

    plt.xlabel("Epoch")

    plt.ylabel("Score")

    plt.title("Precision & Recall")

    plt.legend()

    plt.grid(True)

    plt.show()
````

### Cell 38 — Complete Evaluation Pipeline

````python
def evaluate_pipeline(

    history,

    labels,

    predictions,

    probabilities

):

    metrics = calculate_metrics(

        labels,

        predictions,

        probabilities

    )

    print_metrics(metrics)

    show_classification_report(

        labels,

        predictions

    )

    plot_confusion_matrix(

        labels,

        predictions

    )

    plot_roc_curve(

        labels,

        probabilities

    )

    plot_accuracy(history)

    plot_loss(history)

    plot_precision_recall(history)

    return metrics
````

### Cell 42 — Save Trained Model

````python
def save_model(model):

    save_path = os.path.join(

        MODEL_DIR,

        f"{MODEL_NAME}.keras"

    )

    model.save(save_path)

    print("Model Saved")

    print(save_path)
````

### Cell 43 — Save Training History

````python
def save_history(history):

    history_df = pd.DataFrame(history.history)

    history_path = os.path.join(

        MODEL_DIR,

        f"{MODEL_NAME}_history.csv"

    )

    history_df.to_csv(

        history_path,

        index=False

    )

    print("History Saved")
````

### Cell 44 — Save Performance Metrics

````python
def save_metrics(

    metrics,

    training_time,

    inference_time

):

    metrics_df = pd.DataFrame({

        "Model":[MODEL_NAME],

        "Accuracy":[metrics["Accuracy"]],

        "Precision":[metrics["Precision"]],

        "Recall":[metrics["Recall"]],

        "F1 Score":[metrics["F1 Score"]],

        "ROC AUC":[metrics["ROC AUC"]],

        "Training Time (sec)":[training_time],

        "Inference Time (sec)":[inference_time]

    })

    metrics_path = os.path.join(

        MODEL_DIR,

        f"{MODEL_NAME}_metrics.csv"

    )

    metrics_df.to_csv(

        metrics_path,

        index=False

    )

    print("Metrics Saved")
````

### Cell 45 — Save Confusion Matrix

````python
def save_confusion_matrix(

    labels,

    predictions

):

    cm = confusion_matrix(

        labels,

        predictions

    )

    plt.figure(figsize=(6,5))

    sns.heatmap(

        cm,

        annot=True,

        fmt="d",

        cmap="Blues",

        xticklabels=CLASS_NAMES,

        yticklabels=CLASS_NAMES

    )

    plt.xlabel("Predicted")

    plt.ylabel("Actual")

    plt.title(MODEL_NAME)

    plt.tight_layout()

    plt.savefig(

        os.path.join(

            MODEL_DIR,

            "confusion_matrix.png"

        ),

        dpi=300

    )

    plt.close()
````

### Cell 46 — Save ROC Curve

````python
def save_roc_curve(

    labels,

    probabilities

):

    fpr, tpr, _ = roc_curve(

        labels,

        probabilities

    )

    roc_auc = auc(

        fpr,

        tpr

    )

    plt.figure(figsize=(6,6))

    plt.plot(

        fpr,

        tpr,

        label=f"AUC = {roc_auc:.4f}"

    )

    plt.plot([0,1],[0,1],"--")

    plt.legend()

    plt.grid(True)

    plt.xlabel("False Positive Rate")

    plt.ylabel("True Positive Rate")

    plt.title(MODEL_NAME)

    plt.savefig(

        os.path.join(

            MODEL_DIR,

            "roc_curve.png"

        ),

        dpi=300

    )

    plt.close()
````

### Cell 47 — Save Accuracy & Loss Curves

````python
def save_learning_curves(history):

    # Accuracy

    plt.figure(figsize=(8,5))

    plt.plot(history.history["accuracy"])

    plt.plot(history.history["val_accuracy"])

    plt.legend(["Train","Validation"])

    plt.grid(True)

    plt.savefig(

        os.path.join(

            MODEL_DIR,

            "accuracy_curve.png"

        ),

        dpi=300

    )

    plt.close()

    # Loss

    plt.figure(figsize=(8,5))

    plt.plot(history.history["loss"])

    plt.plot(history.history["val_loss"])

    plt.legend(["Train","Validation"])

    plt.grid(True)

    plt.savefig(

        os.path.join(

            MODEL_DIR,

            "loss_curve.png"

        ),

        dpi=300

    )

    plt.close()
````

### Cell 48 — Save Everything

````python
def save_all(

    model,

    history,

    metrics,

    labels,

    predictions,

    probabilities,

    training_time,

    inference_time

):

    save_model(model)

    save_history(history)

    save_metrics(

        metrics,

        training_time,

        inference_time

    )

    save_confusion_matrix(

        labels,

        predictions

    )

    save_roc_curve(

        labels,

        probabilities

    )

    save_learning_curves(

        history

    )

    print("="*60)

    print("All Files Saved Successfully")

    print("="*60)
````
